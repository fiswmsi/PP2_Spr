```python
#2
import os

path_to_check = r"/Users/sarsenbaisarbinaz/Desktop/pp2/githowto/repositories/w3school/"

def ch(path):
    if not os.path.exists(path):
        print("path doesn't exist")
        return 
    else:
        print('path does exist')
        if os.access(path, os.R_OK):
            print("readable")
        else:
            print("don't readable")

        if os.access(path, os.W_OK):
            print("writable")
        else:
            print("don't writable")
            
        if os.access(path, os.X_OK):
            print("executable")
        else:
            print("don't executable")


    
ch(path_to_check)
```

```python
#5
def ws(el):
    with open("ex.txt", 'w') as f:
        for i in el:
            f.write(str(i) + '\n')
        f.close()
    
 
n = input().split()

ws(n)
```

```python
#6
import os
import string 

def gen():
    for l in string.ascii_uppercase: 
        fn = f"{l}.txt"

        with open(fn, 'w') as file:
            file.write(f"This is {fn}\n")

gen()
```

```python
#7
import os

def cf(A, B):
    with open(A, 'r') as fr, open(B, 'w') as sc:
        sc.write(fr.read())

cf("A.txt", "B.txt")
```

```python
#8
import os

def df(path):
    if os.path.exists(path):  
        if os.access(path, os.W_OK):  
            os.remove(path)
            print(f"File '{path}' has been deleted successfully.")
        else:
            print(f"Cannot delete '{path}'.")
    else:
        print(f"'{path}' does not exist.")

df("D.txt")
```
