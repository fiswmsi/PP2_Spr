```python
#1
from functools import reduce
from operator import mul

def ml(numbers):
    return reduce(mul, numbers)

n = input()
numbers = list(map(int, n.split()))

print(ml(numbers))
```

```python
#2
def count(s):
    uc = sum(1 for char in s if char.isupper())  
    lc = sum(1 for char in s if char.islower())  
    return uc, lc

s = input()
up, low = count(s)

print("Заглавные буквы:", up)
print("Строчные буквы:", low)
```

```python
#3
def pol(s):
    s = s.lower().replace(" ", "") 
    l = ''.join(reversed(s))
    return s == l

s = input()
print(pol(s))
```

```python
#4
import time
import math

num = int(input(""))
num2 = int(input(""))

time.sleep(num2/1000)

sqrt_result = math.sqrt(num)

print(f"Square root of {num} after {num2} milliseconds is {sqrt_result}")
```
