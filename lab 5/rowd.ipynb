```python
import json
import re

def parse_receipt(file_path, output_json):
    with open(file_path, "r", encoding="utf-8") as file:
        raw_text = file.readlines()

    header = {
        "branch": "EUROPHARMA Астана",
        "bin": "080841000761",
        "nds_series": "58001",
        "kassa": "300-189",
        "smena": "10",
        "sequence_number": "64",
        "check_number": "2331180266",
        "kassir": "Аптека 17-1"
    }

    item_pattern = re.compile(r"(\d+)\.\n(.+?)\n([\d\s,]+) x ([\d\s,]+)\n([\d\s,]+)")
    
    items = []
    item_id = 1

    for match in item_pattern.finditer("".join(raw_text)):
        name = match.group(2).strip()
        price = match.group(3).replace(" ", "").strip()
        quantity = match.group(4).replace(" ", "").strip()
        sum_value = match.group(5).replace(" ", "").strip()

        items.append({
            "id": str(item_id),
            "name": name,
            "price": price,
            "quantity": quantity,
            "sum": sum_value
        })
        item_id += 1

    receipt_data = {**header, "items": items}

    with open(output_json, "w", encoding="utf-8") as json_file:
        json.dump(receipt_data, json_file, ensure_ascii=False, indent=4)
    
    return output_json

input_file = "row.txt" 
output_file = "file.json"

parse_receipt(input_file, output_file)
print(f"JSON сохранен в {output_file}")
```
