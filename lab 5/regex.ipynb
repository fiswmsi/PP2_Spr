```python
import re

def tm(text):
        patterns = 'ab*?'
        if re.search(patterns,  text):
                return 'Found a match!'
        else:
                return 'Not matched!'
        
s = input()

print(tm(s))
```

```python
#2

import re

def tm(text):
        patterns = 'ab{2,3}?'
        if re.search(patterns,  text):
                return 'Found a match!'
        else:
                return 'Not matched!'
        
s = input()

print(tm(s))
```

```python
#3

import re

def tm(text):
        patterns = '^[a-z]+_[a-z]+$'
        if re.search(patterns,  text):
                return 'Found a match!'
        else:
                return('Not matched!')
        
s = input()

print(tm(s))
```

```python
#4

import re

def tm(text):
        patterns = '^[A-Z][a-z]+'
        if re.search(patterns,  text):
                return 'Found a match!'
        else:
                return('Not matched!')
        
s = input()

print(tm(s))
```

```python
#5

import re

def tm(text):
        patterns = '^a.*b$'
        if re.search(patterns,  text):
                return 'Found a match!'
        else:
                return('Not matched!')
        
s = input()

print(tm(s))
```
