```python
#1
import math

degree = int(input("Input degree:"))
rad = math.radians(degree)

print(f"Output radian: {rad:.6f}")
```

```python
#2
import math

h = int(input("Height:"))
b1 = int(input("Base, first value:"))
b2 = int(input("Base, second value:"))

area = 0.5 * (b1 + b2) * h

print(f"Expected Output: {area:.1f}")
```

```python
#3
import math

s = int(input("Input number of sides:"))
l = int(input("Input the length of a side: "))
area = (s * l ** 2) / (4 * math.tan(math.pi / s))

print(f"The area of the polygon is: {area:.0f}")
```

```python
#4
import math

l = int(input("Length of base:"))
h = int(input("Height of parallelogram: "))
area = h * l

print(f"Expected Output: {area:.1f}")
```
