```python
#1
def square_generator(N):
    for i in range(N + 1):
        yield i ** 2

N = int(input()) 
for square in square_generator(N):
    print(square)
```

```python
#2
def even_generator(N):
    for i in range(0, N + 1, 2):
        yield i

N = int(input())
n = even_generator(N)

print(",".join(map(str, n)))
```

```python
#3
def generator(N):
    for i in range(0, N + 1):
        if i % 3 == 0 and i % 4 == 0:
            yield i

N = int(input())
n = generator(N)

print(",".join(map(str, n)))
```

```python
#4
def generator(a, b):
    for i in range(a, b + 1):
        yield i ** 2

a = int(input())
b = int(input())

n = generator(a, b)

print(",".join(map(str, n)))
```

```python
#5
def generator(N):
    while N >= 0:
        yield N
        N -=1

N = int(input())

n = generator(N)

print(",".join(map(str, n)))
```
